## Apply

```python
sky = Table().read_table("data/skyscrapers_v2.csv")
sky.show(5)
```

```python
def foot_to_cm(x):
    return 12 * 2.54 * x
```

```python
# Apply a function to every value in a column:
cms = sky.apply(foot_to_cm, 'height')
cms
```

```python
# and add it back to table
sky.with_columns('height in cm', cms)
```

```python
# Create a new table with only one column
Table().with_columns('height in cm', cms)
```

```python
# You can also apply a function with more than one argument.

def bigger(x, y):
    if x > y:
        return x
    else:
        return y    

sky.apply(bigger, 'height', 'completed')
```
